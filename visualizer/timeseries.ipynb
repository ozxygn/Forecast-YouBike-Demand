```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load data
df = pd.read_csv(
    r'C:\Users\asus\Dokumen_Billi\allstationpredgit\result\predictions_all_stations5days.csv',
    parse_dates=['mday']
)

# 2. Pilih satu station
SELECTED_STATION = '500101022'  # <-- ganti dengan station_id yang kamu mau
df_plot = df[df['station_id'] == SELECTED_STATION].sort_values('mday')

# 3. Tentukan list nama kolom model forecast
#    Asumsi kolom actual bernama 'Actual' (sesuaikan kalau header CSV-mu 'actual' lowercase)
actual_col = 'Actual' if 'Actual' in df_plot.columns else 'actual'
model_cols = [c for c in df_plot.columns
              if c not in ['mday', 'station_id', actual_col]]

# 4. Plot actual vs semua model dalam satu figure
plt.figure(figsize=(14, 6))
plt.plot(df_plot['mday'], df_plot[actual_col], label='Actual', linewidth=2)
for model in model_cols:
    plt.plot(df_plot['mday'], df_plot[model], linestyle='--', label=model)

plt.title(f"Actual vs Forecast Models — Station {SELECTED_STATION}")
plt.xlabel("Waktu (mday)")
plt.ylabel("Demand")
plt.legend()
plt.tight_layout()
plt.show()
```
