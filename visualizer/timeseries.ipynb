```python

import pandas as pd
import matplotlib.pyplot as plt

def plot_predictions_for_station(file_path, target_station_id):
    df = pd.read_csv(file_path)
    df["mday"] = pd.to_datetime(df["mday"])

    # Filter for the target station
    station_df = df[df["station_id"] == target_station_id].copy()

    if station_df.empty:
        print(f"No data found for station ID: {target_station_id}")
        return

    # Get all prediction columns
    prediction_cols = [col for col in df.columns if col not in ["mday", "Actual", "station_id"]]

    # Plot Actual Demand
    plt.figure(figsize=(12, 6))
    plt.plot(station_df["mday"], station_df["Actual"], label="Actual Demand", color="blue")
    plt.title(f"Actual Demand for Station {target_station_id}")
    plt.xlabel("Date")
    plt.ylabel("Demand")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"station_{target_station_id}_Actual_Demand.png")
    plt.close()

    # Plot each prediction model separately
    for col in prediction_cols:
        plt.figure(figsize=(12, 6))
        plt.plot(station_df["mday"], station_df["Actual"], label="Actual Demand", color="blue")
        plt.plot(station_df["mday"], station_df[col], label=col, linestyle="--", color="red")
        plt.title(f"Actual vs. {col} Prediction for Station {target_station_id}")
        plt.xlabel("Date")
        plt.ylabel("Demand")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.savefig(f"station_{target_station_id}_{col}_Prediction.png")
        plt.close()

if __name__ == "__main__":
    target_station = 500101022  # User specified station ID
    plot_predictions_for_station(r"C:\Users\asus\Dokumen_Billi\allstationpredgit\result\predictions_all_stations5days.csv", target_station)


```

```python

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.dates as mdates

def plot_predictions_for_station_combined(file_path, target_station_id):
    df = pd.read_excel(file_path)
    df["mday"] = pd.to_datetime(df["mday"])

    # Filter for the target station
    station_df = df[df["station_id"] == target_station_id].copy()

    if station_df.empty:
        print(f"No data found for station ID: {target_station_id}")
        return

    # Get all prediction columns, excluding \'mday\', \'Actual\', and \'station_id\'
    prediction_cols = [col for col in df.columns if col not in ["mday", "Actual", "station_id"]]

    plt.figure(figsize=(15, 8))
    plt.plot(station_df["mday"], station_df["Actual"], label="Actual Demand", color="black", linewidth=2)

    # Generate a colormap for different prediction models
    colors = cm.get_cmap("tab10", len(prediction_cols))

    for i, col in enumerate(prediction_cols):
        plt.plot(station_df["mday"], station_df[col], label=col, linestyle="-", color=colors(i)) # Changed linestyle to solid

    plt.title(f"Actual vs. All Prediction Models for Station {target_station_id}")
    plt.xlabel("Time of Day")
    plt.ylabel("Available Rent Bikes") # Changed y-axis label
    plt.legend(loc="upper left", bbox_to_anchor=(1, 1))
    plt.grid(True)

    # Format x-axis to show only hours
    formatter = mdates.DateFormatter("%H:%M")
    plt.gca().xaxis.set_major_formatter(formatter)

    plt.tight_layout()
    plt.savefig(f"station_{target_station_id}_All_Predictions_Combined_Hourly_RentBikes.png")
    plt.close()

if __name__ == "__main__":
    target_station = 500101022  # User specified station ID
    plot_predictions_for_station_combined("C:\\Users\\asus\\Dokumen_Billi\\allstationpredgit\\Prediction_all_model.xlsx", target_station)


```
